```python
# ============================================================
# STEP 7: CREATE AGRIBUSINESS DATASET
# ============================================================

import pandas as pd
import numpy as np

np.random.seed(42)

states = [
    "Punjab", "Haryana", "Uttar Pradesh", "Madhya Pradesh",
    "Maharashtra", "Rajasthan", "Bihar", "West Bengal"
]

crops = [
    "Wheat", "Rice", "Maize", "Cotton", "Sugarcane",
    "Potato", "Soybean", "Pulses"
]

seasons = ["Kharif", "Rabi", "Summer"]

years = list(range(2018, 2024))

data = []

for year in years:
    for state in states:
        for crop in crops:

            area = np.random.randint(5000, 100000)

            base_yield = {
                "Wheat": 3.2,
                "Rice": 2.8,
                "Maize": 2.5,
                "Cotton": 1.8,
                "Sugarcane": 70,
                "Potato": 22,
                "Soybean": 1.4,
                "Pulses": 1.1
            }[crop]

            yield_value = round(
                base_yield * np.random.uniform(0.85, 1.15), 2
            )

            production = round(area * yield_value, 2)

            rainfall = round(np.random.uniform(450, 1200), 1)
            temperature = round(np.random.uniform(18, 34), 1)

            price = round(
                np.random.uniform(1500, 6000), 2
            )

            revenue = round(production * price, 2)

            season = np.random.choice(seasons)

            data.append([
                state,
                crop,
                year,
                season,
                area,
                production,
                yield_value,
                rainfall,
                temperature,
                price,
                revenue
            ])

columns = [
    "State",
    "Crop",
    "Crop_Year",
    "Season",
    "Area_Hectares",
    "Production_Tonnes",
    "Yield_Tonnes_per_Hectare",
    "Rainfall_mm",
    "Temperature_C",
    "Market_Price",
    "Revenue"
]

df = pd.DataFrame(data, columns=columns)

print("Agribusiness dataset created successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

display(df.head())
```

```python
# ============================================================
# STEP 13: OUTLIER DETECTION USING IQR
# ============================================================

numeric_columns = [
    "Area_Hectares",
    "Production_Tonnes",
    "Yield_Tonnes_per_Hectare",
    "Rainfall_mm",
    "Temperature_C",
    "Market_Price",
    "Revenue"
]

outlier_summary = []

for column in numeric_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    outlier_summary.append([
        column,
        len(outliers),
        round(lower_limit, 2),
        round(upper_limit, 2)
    ])

outlier_df = pd.DataFrame(
    outlier_summary,
    columns=[
        "Column",
        "Outlier_Count",
        "Lower_Limit",
        "Upper_Limit"
    ]
)

display(outlier_df)
```

```python
# ============================================================
# STEP 14: DATA CLEANING AND VALIDATION
# ============================================================

# Create a copy so the original dataset remains protected
cleaned_df = df.copy()

# ------------------------------------------------------------
# 1. Remove duplicate records
# ------------------------------------------------------------

cleaned_df = cleaned_df.drop_duplicates()

# ------------------------------------------------------------
# 2. Clean text columns
# ------------------------------------------------------------

text_columns = ["State", "Crop", "Season"]

for column in text_columns:
    cleaned_df[column] = (
        cleaned_df[column]
        .astype(str)
        .str.strip()
        .str.title()
    )

# ------------------------------------------------------------
# 3. Convert numeric columns to numeric data type
# ------------------------------------------------------------

numeric_columns = [
    "Crop_Year",
    "Area_Hectares",
    "Production_Tonnes",
    "Yield_Tonnes_per_Hectare",
    "Rainfall_mm",
    "Temperature_C",
    "Market_Price",
    "Revenue"
]

for column in numeric_columns:
    cleaned_df[column] = pd.to_numeric(
        cleaned_df[column],
        errors="coerce"
    )

# ------------------------------------------------------------
# 4. Check missing values after conversion
# ------------------------------------------------------------

print("Missing Values After Cleaning:")
display(
    cleaned_df.isnull().sum().to_frame(
        name="Missing_Values"
    )
)

# ------------------------------------------------------------
# 5. Check invalid negative values
# ------------------------------------------------------------

positive_columns = [
    "Area_Hectares",
    "Production_Tonnes",
    "Yield_Tonnes_per_Hectare",
    "Rainfall_mm",
    "Market_Price",
    "Revenue"
]

print("\nNegative Values Check:")
print("=" * 50)

for column in positive_columns:
    negative_count = (cleaned_df[column] < 0).sum()
    print(f"{column}: {negative_count}")

# ------------------------------------------------------------
# 6. Final dataset information
# ------------------------------------------------------------

print("\nFinal Dataset Information:")
print("=" * 50)

print("Rows:", cleaned_df.shape[0])
print("Columns:", cleaned_df.shape[1])

print("\nDuplicate Records:",
      cleaned_df.duplicated().sum())

print("\nTotal Missing Values:",
      cleaned_df.isnull().sum().sum())
```

```python
# ============================================================
# WEEK 3 - STEP 6: LINEAR REGRESSION - FIXED
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    mean_absolute_percentage_error,
    r2_score
)

# ------------------------------------------------------------
# RECOVER DATA IF NECESSARY
# ------------------------------------------------------------

if "cleaned_df" not in globals():

    print("cleaned_df not found. Recreating dataset...")

    np.random.seed(42)

    states = [
        "Punjab", "Haryana", "Uttar Pradesh", "Madhya Pradesh",
        "Maharashtra", "Rajasthan", "Bihar", "West Bengal"
    ]

    crops = [
        "Wheat", "Rice", "Maize", "Cotton",
        "Sugarcane", "Potato", "Soybean", "Pulses"
    ]

    seasons = ["Kharif", "Rabi", "Summer"]

    data = []

    for year in range(2018, 2024):
        for state in states:
            for crop in crops:

                area = np.random.randint(5000, 100000)

                base_yield = {
                    "Wheat": 3.2,
                    "Rice": 2.8,
                    "Maize": 2.5,
                    "Cotton": 1.8,
                    "Sugarcane": 70,
                    "Potato": 22,
                    "Soybean": 1.4,
                    "Pulses": 1.1
                }[crop]

                yield_value = round(
                    base_yield * np.random.uniform(0.85, 1.15),
                    2
                )

                production = round(
                    area * yield_value,
                    2
                )

                rainfall = round(
                    np.random.uniform(450, 1200),
                    1
                )

                temperature = round(
                    np.random.uniform(18, 34),
                    1
                )

                price = round(
                    np.random.uniform(1500, 6000),
                    2
                )

                revenue = round(
                    production * price,
                    2
                )

                season = np.random.choice(seasons)

                data.append([
                    state, crop, year, season,
                    area, production, yield_value,
                    rainfall, temperature, price, revenue
                ])

    columns = [
        "State",
        "Crop",
        "Crop_Year",
        "Season",
        "Area_Hectares",
        "Production_Tonnes",
        "Yield_Tonnes_per_Hectare",
        "Rainfall_mm",
        "Temperature_C",
        "Market_Price",
        "Revenue"
    ]

    cleaned_df = pd.DataFrame(data, columns=columns)


# ------------------------------------------------------------
# FEATURES AND TARGET
# ------------------------------------------------------------

features = [
    "Area_Hectares",
    "Production_Tonnes",
    "Rainfall_mm",
    "Temperature_C",
    "Market_Price",
    "Crop_Year"
]

target = "Yield_Tonnes_per_Hectare"


# ------------------------------------------------------------
# CHRONOLOGICAL TRAIN / TEST SPLIT
# ------------------------------------------------------------

model_df = cleaned_df.sort_values(
    "Crop_Year"
).reset_index(drop=True)

split_index = int(len(model_df) * 0.80)

train_df = model_df.iloc[:split_index]
test_df = model_df.iloc[split_index:]

X_train = train_df[features]
X_test = test_df[features]

y_train = train_df[target]
y_test = test_df[target]


# ------------------------------------------------------------
# TRAIN MODEL
# ------------------------------------------------------------

linear_model = LinearRegression()

linear_model.fit(
    X_train,
    y_train
)


# ------------------------------------------------------------
# PREDICTIONS
# ------------------------------------------------------------

linear_predictions = linear_model.predict(
    X_test
)


# ------------------------------------------------------------
# EVALUATION
# ------------------------------------------------------------

linear_mae = mean_absolute_error(
    y_test,
    linear_predictions
)

linear_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        linear_predictions
    )
)

linear_mape = mean_absolute_percentage_error(
    y_test,
    linear_predictions
) * 100

linear_r2 = r2_score(
    y_test,
    linear_predictions
)


# ------------------------------------------------------------
# RESULTS
# ------------------------------------------------------------

print("=" * 65)
print("       MULTIPLE LINEAR REGRESSION RESULTS")
print("=" * 65)

print(f"Training Records : {len(train_df)}")
print(f"Testing Records  : {len(test_df)}")

print(f"\nMAE  : {linear_mae:.4f}")
print(f"RMSE : {linear_rmse:.4f}")
print(f"MAPE : {linear_mape:.2f}%")
print(f"R²   : {linear_r2:.4f}")


# ------------------------------------------------------------
# ACTUAL VS PREDICTED
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

plt.scatter(
    y_test,
    linear_predictions,
    alpha=0.6
)

min_value = min(
    y_test.min(),
    linear_predictions.min()
)

max_value = max(
    y_test.max(),
    linear_predictions.max()
)

plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    linestyle="--"
)

plt.title(
    "Actual vs Predicted Crop Yield - Linear Regression"
)

plt.xlabel(
    "Actual Yield (Tonnes/Hectare)"
)

plt.ylabel(
    "Predicted Yield (Tonnes/Hectare)"
)

plt.grid(True, alpha=0.3)

plt.tight_layout()

plt.savefig(
    "linear_regression_actual_vs_predicted.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nLinear Regression completed successfully!")
```

```python
# ============================================================
# WEEK 3 - STEP 12
# MAIZE YIELD FORECASTING - FIXED VERSION
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. RECOVER / CREATE DATASET
# ------------------------------------------------------------

if "cleaned_df" not in globals():

    print("cleaned_df not found. Recreating dataset...")

    np.random.seed(42)

    states = [
        "Punjab", "Haryana", "Uttar Pradesh", "Madhya Pradesh",
        "Maharashtra", "Rajasthan", "Bihar", "West Bengal"
    ]

    crops = [
        "Wheat", "Rice", "Maize", "Cotton",
        "Sugarcane", "Potato", "Soybean", "Pulses"
    ]

    seasons = ["Kharif", "Rabi", "Summer"]

    data = []

    for year in range(2018, 2024):

        for state in states:

            for crop in crops:

                area = np.random.randint(5000, 100000)

                base_yield = {
                    "Wheat": 3.2,
                    "Rice": 2.8,
                    "Maize": 2.5,
                    "Cotton": 1.8,
                    "Sugarcane": 70,
                    "Potato": 22,
                    "Soybean": 1.4,
                    "Pulses": 1.1
                }[crop]

                yield_value = round(
                    base_yield * np.random.uniform(0.85, 1.15),
                    2
                )

                production = round(
                    area * yield_value,
                    2
                )

                rainfall = round(
                    np.random.uniform(450, 1200),
                    1
                )

                temperature = round(
                    np.random.uniform(18, 34),
                    1
                )

                price = round(
                    np.random.uniform(1500, 6000),
                    2
                )

                revenue = round(
                    production * price,
                    2
                )

                season = np.random.choice(seasons)

                data.append([
                    state,
                    crop,
                    year,
                    season,
                    area,
                    production,
                    yield_value,
                    rainfall,
                    temperature,
                    price,
                    revenue
                ])

    columns = [
        "State",
        "Crop",
        "Crop_Year",
        "Season",
        "Area_Hectares",
        "Production_Tonnes",
        "Yield_Tonnes_per_Hectare",
        "Rainfall_mm",
        "Temperature_C",
        "Market_Price",
        "Revenue"
    ]

    cleaned_df = pd.DataFrame(
        data,
        columns=columns
    )

    cleaned_df = cleaned_df.drop_duplicates()

# ------------------------------------------------------------
# 2. SELECT MAIZE
# ------------------------------------------------------------

maize_df = cleaned_df[
    cleaned_df["Crop"].str.lower() == "maize"
].copy()

print("Maize records:", len(maize_df))

# ------------------------------------------------------------
# 3. YEAR-WISE MAIZE YIELD
# ------------------------------------------------------------

maize_yearly = (
    maize_df
    .groupby("Crop_Year")[
        "Yield_Tonnes_per_Hectare"
    ]
    .mean()
    .sort_index()
)

print("\n" + "=" * 70)
print("             HISTORICAL MAIZE YIELD")
print("=" * 70)

display(
    maize_yearly
    .to_frame("Average_Yield")
    .round(3)
)

# ------------------------------------------------------------
# 4. TREND-BASED FORECAST
# ------------------------------------------------------------

years = maize_yearly.index.values
yields = maize_yearly.values

coefficients = np.polyfit(
    years,
    yields,
    1
)

trend_model = np.poly1d(
    coefficients
)

future_years = np.arange(
    years.max() + 1,
    years.max() + 4
)

future_predictions = trend_model(
    future_years
)

# Prevent negative forecast values
future_predictions = np.maximum(
    future_predictions,
    0
)

forecast_table = pd.DataFrame({
    "Year": future_years,
    "Forecast_Maize_Yield": future_predictions
})

print("\n" + "=" * 70)
print("             MAIZE YIELD FORECAST")
print("=" * 70)

display(
    forecast_table.round(3)
)

# ------------------------------------------------------------
# 5. HISTORICAL + FORECAST VISUALIZATION
# ------------------------------------------------------------

plt.figure(figsize=(11, 6))

plt.plot(
    years,
    yields,
    marker="o",
    label="Historical Yield"
)

plt.plot(
    future_years,
    future_predictions,
    marker="o",
    linestyle="--",
    label="Forecast Yield"
)

plt.title(
    "Maize Yield: Historical Trend and Forecast"
)

plt.xlabel("Year")

plt.ylabel(
    "Average Yield (Tonnes/Hectare)"
)

plt.legend()

plt.grid(
    True,
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    "maize_yield_forecast.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nMaize yield forecasting scenario completed successfully!")
```

```python
# ============================================================
# WEEK 4 - STEP 4
# KPI PRIORITY RANKING
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt

# Check KPI dataset
if "kpi_data" not in globals():

    print("kpi_data not found.")
    print("Please run Week 4 - Step 1 first.")

else:

    # --------------------------------------------------------
    # Calculate gap if not already available
    # --------------------------------------------------------

    kpi_data["Gap"] = (
        kpi_data["Actual_Performance"]
        - kpi_data["Benchmark"]
    )

    # --------------------------------------------------------
    # Priority based on negative gap
    # --------------------------------------------------------

    kpi_priority = kpi_data.copy()

    kpi_priority["Priority_Score"] = (
        -kpi_priority["Gap"]
    )

    kpi_priority = kpi_priority.sort_values(
        "Priority_Score",
        ascending=False
    ).reset_index(drop=True)

    kpi_priority["Priority_Rank"] = (
        kpi_priority.index + 1
    )

    # --------------------------------------------------------
    # Priority category
    # --------------------------------------------------------

    kpi_priority["Priority"] = "Low"

    kpi_priority.loc[
        kpi_priority["Gap"] < -10,
        "Priority"
    ] = "High"

    kpi_priority.loc[
        (kpi_priority["Gap"] >= -10) &
        (kpi_priority["Gap"] < 0),
        "Priority"
    ] = "Medium"

    kpi_priority.loc[
        kpi_priority["Gap"] >= 0,
        "Priority"
    ] = "Low"

    # --------------------------------------------------------
    # Display
    # --------------------------------------------------------

    print("=" * 80)
    print("                 KPI PRIORITY RANKING")
    print("=" * 80)

    display(
        kpi_priority[
            [
                "Priority_Rank",
                "KPI",
                "Actual_Performance",
                "Benchmark",
                "Gap",
                "Priority"
            ]
        ].round(2)
    )

    # --------------------------------------------------------
    # Priority visualization
    # --------------------------------------------------------

    plt.figure(figsize=(11, 6))

    plt.barh(
        kpi_priority["KPI"],
        kpi_priority["Gap"]
    )

    plt.axvline(
        0,
        linestyle="--"
    )

    plt.title(
        "KPI Performance Gap and Improvement Priority"
    )

    plt.xlabel(
        "Performance Gap (Actual - Benchmark)"
    )

    plt.ylabel("KPI")

    plt.tight_layout()

    plt.savefig(
        "kpi_priority_ranking.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    print("\nKPI priority ranking completed successfully!")
```

```python
# ============================================================
# WEEK 4 - STEP 21
# INTELLIGENT KPI RISK & RECOMMENDATION ENGINE
# ============================================================

import pandas as pd
import numpy as np

# Check KPI dataset
if "kpi_data" not in globals():

    print("kpi_data not found.")
    print("Please run the previous Week 4 steps first.")

else:

    # --------------------------------------------------------
    # Create working dataset
    # --------------------------------------------------------

    intelligent_kpi = kpi_data.copy()

    # --------------------------------------------------------
    # Calculate performance metrics
    # --------------------------------------------------------

    intelligent_kpi["Gap"] = (
        intelligent_kpi["Actual_Performance"]
        - intelligent_kpi["Benchmark"]
    )

    intelligent_kpi["Performance_Score"] = (
        intelligent_kpi["Actual_Performance"]
        / intelligent_kpi["Benchmark"]
    ) * 100

    intelligent_kpi["Gap_Percentage"] = (
        intelligent_kpi["Gap"]
        / intelligent_kpi["Benchmark"]
    ) * 100

    # --------------------------------------------------------
    # Risk Score
    #
    # Negative gap = higher risk
    # Positive gap = lower risk
    # --------------------------------------------------------

    intelligent_kpi["Risk_Score"] = np.where(
        intelligent_kpi["Gap"] < 0,
        abs(intelligent_kpi["Gap_Percentage"]),
        0
    )

    # --------------------------------------------------------
    # Risk Category
    # --------------------------------------------------------

    def classify_risk(score):

        if score >= 15:
            return "High Risk"

        elif score >= 5:
            return "Medium Risk"

        else:
            return "Low Risk"

    intelligent_kpi["Risk_Category"] = (
        intelligent_kpi["Risk_Score"]
        .apply(classify_risk)
    )

    # --------------------------------------------------------
    # Intelligent recommendations
    # --------------------------------------------------------

    def generate_recommendation(row):

        kpi = row["KPI"]
        risk = row["Risk_Category"]

        if kpi == "Price Risk Coverage":
            return (
                "Prioritize price-risk management using "
                "forward contracts, hedging and market planning."
            )

        elif kpi == "Cost per Ton Produced":
            return (
                "Optimize procurement, input usage and "
                "operational costs to reduce production cost."
            )

        elif kpi == "Water Use Efficiency":
            return (
                "Improve irrigation scheduling, monitoring "
                "and water-efficient resource management."
            )

        elif kpi == "Gross Profit Margin":
            return (
                "Improve profitability through cost control "
                "and stronger pricing strategies."
            )

        elif kpi == "Yield per Hectare":
            return (
                "Maintain the existing yield advantage through "
                "productive agricultural practices."
            )

        else:
            return (
                "Monitor the KPI and develop a targeted "
                "improvement strategy."
            )

    intelligent_kpi["Intelligent_Recommendation"] = (
        intelligent_kpi.apply(
            generate_recommendation,
            axis=1
        )
    )

    # --------------------------------------------------------
    # Sort by risk
    # --------------------------------------------------------

    intelligent_kpi = intelligent_kpi.sort_values(
        "Risk_Score",
        ascending=False
    ).reset_index(drop=True)

    intelligent_kpi["AI_Priority_Rank"] = (
        intelligent_kpi.index + 1
    )

    # --------------------------------------------------------
    # Display
    # --------------------------------------------------------

    print("=" * 110)
    print("        INTELLIGENT KPI RISK & RECOMMENDATION ENGINE")
    print("=" * 110)

    display(
        intelligent_kpi[
            [
                "AI_Priority_Rank",
                "KPI",
                "Performance_Score",
                "Gap",
                "Risk_Score",
                "Risk_Category",
                "Intelligent_Recommendation"
            ]
        ].round(2)
    )

    # --------------------------------------------------------
    # Highest-risk KPI
    # --------------------------------------------------------

    highest_risk = intelligent_kpi.iloc[0]

    print("\n" + "=" * 80)
    print("                 INTELLIGENT PRIORITY ALERT")
    print("=" * 80)

    print(
        f"\nHighest Risk KPI : {highest_risk['KPI']}"
    )

    print(
        f"Risk Score       : "
        f"{highest_risk['Risk_Score']:.2f}"
    )

    print(
        f"Risk Category    : "
        f"{highest_risk['Risk_Category']}"
    )

    print(
        f"\nRecommended Action:\n"
        f"{highest_risk['Intelligent_Recommendation']}"
    )

    # --------------------------------------------------------
    # Save intelligent analysis
    # --------------------------------------------------------

    intelligent_kpi.to_csv(
        "Week4_Intelligent_KPI_Risk_Analysis.csv",
        index=False
    )

    print(
        "\nIntelligent KPI Risk Analysis completed successfully!"
    )

    print(
        "Saved as: Week4_Intelligent_KPI_Risk_Analysis.csv"
    )
```

```python

# ============================================================
# WEEK 4 - STEP 27
# HISTORICAL AGRIBUSINESS KPI DATASET FOR ML
# ============================================================

import pandas as pd
import numpy as np

# Reproducibility
np.random.seed(42)

# ------------------------------------------------------------
# Generate historical observations
# ------------------------------------------------------------

n = 120

historical_data = pd.DataFrame({

    "Yield_per_Hectare":
        np.random.normal(104, 7, n),

    "Cost_per_Ton":
        np.random.normal(90, 6, n),

    "Gross_Profit_Margin":
        np.random.normal(95, 5, n),

    "Water_Use_Efficiency":
        np.random.normal(90, 6, n),

    "Price_Risk_Coverage":
        np.random.normal(80, 8, n)
})

# ------------------------------------------------------------
# Keep values within realistic ranges
# ------------------------------------------------------------

historical_data["Yield_per_Hectare"] = (
    historical_data["Yield_per_Hectare"].clip(70, 130)
)

historical_data["Cost_per_Ton"] = (
    historical_data["Cost_per_Ton"].clip(60, 120)
)

historical_data["Gross_Profit_Margin"] = (
    historical_data["Gross_Profit_Margin"].clip(70, 120)
)

historical_data["Water_Use_Efficiency"] = (
    historical_data["Water_Use_Efficiency"].clip(65, 120)
)

historical_data["Price_Risk_Coverage"] = (
    historical_data["Price_Risk_Coverage"].clip(50, 110)
)

# ------------------------------------------------------------
# Create overall performance target
# ------------------------------------------------------------

historical_data["Overall_Performance"] = (
    0.30 * historical_data["Yield_per_Hectare"]
    +
    0.20 * historical_data["Cost_per_Ton"]
    +
    0.20 * historical_data["Gross_Profit_Margin"]
    +
    0.15 * historical_data["Water_Use_Efficiency"]
    +
    0.15 * historical_data["Price_Risk_Coverage"]
)

# Add small natural variation
historical_data["Overall_Performance"] += (
    np.random.normal(0, 2, n)
)

# ------------------------------------------------------------
# Display dataset
# ------------------------------------------------------------

print("=" * 90)
print("       HISTORICAL AGRIBUSINESS KPI DATASET")
print("=" * 90)

display(
    historical_data.head(10).round(2)
)

print("\nDataset Shape:", historical_data.shape)

print(
    "\nHistorical ML dataset generated successfully!"
)

# ------------------------------------------------------------
# Save dataset
# ------------------------------------------------------------

historical_data.to_csv(
    "Agribusiness_Historical_KPI_ML_Dataset.csv",
    index=False
)

print(
    "Saved as: Agribusiness_Historical_KPI_ML_Dataset.csv"
)
```

```python
# ============================================================
# WEEK 4 - STEP 28
# ML DATA PREPARATION
# ============================================================

from sklearn.model_selection import train_test_split

# Check dataset
if "historical_data" not in globals():

    print("historical_data not found.")
    print("Please run Week 4 - Step 27 first.")

else:

    # --------------------------------------------------------
    # Define features and target
    # --------------------------------------------------------

    features = [
        "Yield_per_Hectare",
        "Cost_per_Ton",
        "Gross_Profit_Margin",
        "Water_Use_Efficiency",
        "Price_Risk_Coverage"
    ]

    target = "Overall_Performance"

    X = historical_data[features]
    y = historical_data[target]

    # --------------------------------------------------------
    # Train-test split
    # --------------------------------------------------------

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=42
    )

    # --------------------------------------------------------
    # Display information
    # --------------------------------------------------------

    print("=" * 80)
    print("              ML DATA PREPARATION")
    print("=" * 80)

    print(f"\nTotal observations : {len(historical_data)}")
    print(f"Training samples   : {len(X_train)}")
    print(f"Testing samples    : {len(X_test)}")
    print(f"Number of features : {len(features)}")

    print("\nFeatures:")
    for feature in features:
        print(f"• {feature}")

    print(f"\nTarget variable: {target}")

    print("\nTrain-test split completed successfully!")
```

```python
# ============================================================
# WEEK 4 - STEP 29
# RANDOM FOREST REGRESSION MODEL
# ============================================================

from sklearn.ensemble import RandomForestRegressor

# Check ML data
if "X_train" not in globals():

    print("Training data not found.")
    print("Please run Week 4 - Step 28 first.")

else:

    # --------------------------------------------------------
    # Create Random Forest model
    # --------------------------------------------------------

    rf_model = RandomForestRegressor(
        n_estimators=200,
        max_depth=8,
        min_samples_split=4,
        random_state=42
    )

    # --------------------------------------------------------
    # Train model
    # --------------------------------------------------------

    rf_model.fit(
        X_train,
        y_train
    )

    # --------------------------------------------------------
    # Generate predictions
    # --------------------------------------------------------

    y_pred = rf_model.predict(
        X_test
    )

    # --------------------------------------------------------
    # Display sample predictions
    # --------------------------------------------------------

    prediction_comparison = pd.DataFrame({

        "Actual_Performance":
            y_test.values,

        "Predicted_Performance":
            y_pred

    })

    print("=" * 90)
    print("          RANDOM FOREST REGRESSION MODEL")
    print("=" * 90)

    print("\nModel training completed successfully!")

    print("\nSample Predictions:")

    display(
        prediction_comparison
        .head(10)
        .round(2)
    )
```

```python
# ============================================================
# WEEK 4 - STEP 30
# MACHINE LEARNING MODEL EVALUATION
# ============================================================

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
import numpy as np

# Check model predictions
if "y_pred" not in globals():

    print("y_pred not found.")
    print("Please run Week 4 - Step 29 first.")

else:

    # --------------------------------------------------------
    # Calculate evaluation metrics
    # --------------------------------------------------------

    mae = mean_absolute_error(
        y_test,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_test,
            y_pred
        )
    )

    r2 = r2_score(
        y_test,
        y_pred
    )

    # --------------------------------------------------------
    # Display metrics
    # --------------------------------------------------------

    print("=" * 80)
    print("          MACHINE LEARNING MODEL EVALUATION")
    print("=" * 80)

    print(
        f"\nMean Absolute Error (MAE) : {mae:.4f}"
    )

    print(
        f"Root Mean Squared Error   : {rmse:.4f}"
    )

    print(
        f"R² Score                  : {r2:.4f}"
    )

    # --------------------------------------------------------
    # Interpretation
    # --------------------------------------------------------

    print("\nMODEL INTERPRETATION")
    print("-" * 60)

    if r2 >= 0.80:

        print(
            "The model demonstrates strong predictive performance."
        )

    elif r2 >= 0.60:

        print(
            "The model demonstrates moderate predictive performance."
        )

    else:

        print(
            "The model requires further improvement or more data."
        )

    print(
        "\nMachine learning model evaluation completed successfully!"
    )
```
